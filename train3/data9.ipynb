```python

titanic.head()
```

```python
sns.barplot(
    x=titanic['Pclass'],
    y=titanic['Fare'],
    hue=titanic['Sex']
)
```

```python
sns.barplot(
    x=titanic['Pclass'],
    y=titanic['Age'],
    hue=titanic['Sex']
)
```

```python
sns.boxplot(
    x='Sex',
    y='Age',
    hue='Survived',
    data=titanic
)
```

```python
sns.kdeplot(
    data=titanic[titanic['Survived'] == 0],
    x='Age',
    fill=False,
    label='Did Not Survive'
)

sns.kdeplot(
    data=titanic[titanic['Survived'] == 1],
    x='Age',
    fill=False,
    label='Survived'
)
```

```python
titanic.head(3)
```

```python
pd.crosstab(titanic['Pclass'],titanic['Survived'])
```

```python
sns.heatmap(pd.crosstab(titanic['Pclass'],titanic['Survived']))
```

```python
(titanic.groupby('Pclass')['Survived']
        .mean()
        .mul(100)
        .round(2))
```

```python
(titanic.groupby('Pclass')['Survived']
        .mean()
        .mul(100)
        .round(2)).plot(kind='bar')
```

```python
(titanic.groupby('Embarked')['Survived']
        .mean()
        .mul(100)
        .round(2))
```

```python
(titanic.groupby('Embarked')['Survived']
        .mean()
        .mul(100)
        .round(2)).plot(kind='bar')
```

```python
(titanic.groupby('Sex')['Survived']
        .mean()
        .mul(100)
        .round(2))
```

```python
(titanic.groupby('Sex')['Survived']
        .mean()
        .mul(100)
        .round(2)).plot(kind='bar')
```

```python
pd.crosstab(titanic['Parch'],titanic['Survived'])
```

```python
sns.clustermap(pd.crosstab(titanic['Parch'],titanic['Survived']))
```
